```python
import seaborn as sns
import pandas as pd
```

```python
penguins = pd.read_csv('Dataset/penguins.csv')
```

```python
penguins
```

Panjang Paruh vs Lebar Paruh

```python
sns.scatterplot(
    data = penguins,
    x = 'bill_length_mm',
    y = 'bill_depth_mm'
)
```

ternyata panjang paruh ga berpengaruh banyak ke lebar paruh

---

Panjang Paruh vs Panjang Sayap

```python
sns.scatterplot(
    data = penguins,
    x = 'bill_length_mm',
    y = 'flipper_length_mm'
)
```

Ternyata ada keterkaitan antara panjang paruh sama berat badan.

---

Panjang Paruh vs Berat Badan

```python
sns.scatterplot(
    data = penguins,
    x = 'bill_length_mm',
    y = 'body_mass_g'
)
```

ada sedikit keterkaitan antara panjang paruh dengan berat badang penguins

---

Panjang Sayap vs Berat Badan

```python
sns.scatterplot(
    data = penguins,
    x = 'flipper_length_mm',
    y = 'body_mass_g'
)
```
