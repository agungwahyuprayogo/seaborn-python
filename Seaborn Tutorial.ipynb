# set_context()

Fungsi ini digunakan untuk **mengatur skala elemen visual (seperti ukuran label, garis, dan titik) agar sesuai dengan media** di mana grafik tersebut akan ditampilkan.


| Context | Ukuran | Deskripsi |
| --- | --- | --- |
| **`"paper"`** | Elemen terkecil | Ideal untuk dicetak dalam karya ilmiah atau jurnal akademik. |
| **`"notebook"`** | Ukuran sedang | Cocok untuk bekerja di dalam Jupyter Notebook (pengaturan standar/default). |
| **`"talk"`** | Teks dan garis lebih besar | Bagus untuk slide presentasi di mana keterbacaan sangat diutamakan. |
| **`"poster"`** | Elemen sangat besar | Cocok untuk tampilan layar besar atau poster fisik. |

## Berdasarkan Jenis Data

**Panduan Palet Warna Seaborn**

| Jenis Palet | Tujuan | Contoh | Penggunaan Terbaik |
| --- | --- | --- | --- |
| **Kualitatif (*Qualitative*)** | Data yang bersifat kategorikal; warna yang diberikan pada setiap grup harus terlihat berbeda secara jelas. | `'Set1'`, `'Set2'`, `'Set3'`, `'Pastel1'`, `'Dark2'`, `'Colorblind'`, `'Accent'` | Membedakan kategori atau grup (misal: nama stasiun, jenis polutan). |
| **Sekuensial (*Sequential*)** | Data dengan urutan atau perkembangan yang jelas (rendah -> tinggi). | `'Blues'`, `'Greens'`, `'Purples'`, `'OrRd'`, `'YlGnBu'`, `'BuPu'`, `'Greys'` | Menunjukkan besaran atau intensitas (misal: jumlah populasi, pendapatan, tingkat PM10). |
| **Divergen (*Diverging*)** | Data yang memiliki titik tengah yang bermakna (misal: angka 0) atau nilai campuran seperti positif dan negatif. | `'coolwarm'`, `'RdBu'`, `'BrBG'`, `'PiYG'`, `'Spectral'`, `'terrain_r'` | Menunjukkan perubahan positif vs negatif (misal: laba/rugi, korelasi, perubahan suhu). |

---

Jika nanti kamu menemukan nama palet seperti **"YlOrRd"**, ingat kuncinya:

- **Yl** = Yellow
- **Or** = Orange
- **Rd** = Red Semuanya menggunakan huruf L kecil atau R kecil, bukan huruf besar di tengah kata.

# Box plot

oke, selanjutnya kita akan menggunakan `box plot`, dimana **plot ini digunakan untuk mendeteksi `outlier`** yang terlalu rendah atau terlalu tinggi di sebuah fitur atau kolom. Jadi pada dasarnya `box plot`membantu kita mengidentifikasi kelompok-kelompok yang berbeda dalam suatu fitur melalui kuartilnya.

Secara umum `box plot` menggunakan 3 kuartil, yaitu 25%, 50%, dan 75%. Jadi bisa kita lihat dari gambar di bawah :

Jadi bentuk dari box pot tuh kaya diatas, kalian bisa lihat titik yang paling bawah itu `outlier`. **`Outlier` itu atau datanya utlier (pencilan) adalah titik data yang nilainya berbeda secara signifikan dari sebagian besar data lainnya dalam satu kumpulan data.** 

Bayangkan kamu sedang mendata tinggi badan siswa **di sebuah kelas yang rata-ratanya 160−170 cm**, tiba-tiba **ada satu orang yang tingginya 210 cm.** Nah, **orang dengan tinggi 210 cm itulah yang disebut `outlier`.**

Dan biasanya **box plot memiliki *batas maksimum* atau `upper extreme` dan juga *batas minimum* atau `lower extreme`.**

- Kalo untuk contoh gambar diatas, batas maksimumnya adalah 90.
- batas minimumnya 10
- data outlier-nya adalah 3
- Median atau nilai tengahnya adalah 55
- Lalu kalo untuk kotak disekitar median itu disebut quartile,
    - garis tepi kotak yang bawah itu quartile 25% atau bisa di sebut lower quartile
    - dan garis tepi kotak yang atas itu quartile 75% upper quartile
    - sedangkan median itu yang di garis tengah-tengah kotak tersebut

Oke, jadi langsung coba saja menggunakan boxplot :

````python
sns.boxplot(data = penguins, x='species', y = 'body_mass_g', hue='island')
````

Oke, jadi disini kita punya 3 species berbeda. Nah untuk warna itu mewakili pulaunya. 

Kita juga bisa mengelompokan berdasarkan kelamin :

````python
sns.boxplot(data = penguins, x='species', y = 'body_mass_g', hue='sex')
````

Kalo kita lihat sekilas, species **`Chinstrap` punya penguins yang `outlier`** diatas dan bawah. Untuk yang **diatas bisa dibilang kaya *obesitas***, dan **`outlier` yang bawah, mungkin kalo kita nyebutnya tuh kekurangan gizi** :'v

Lalu, kita bisa melihat median dari tiap-tiap kelamin species : 

- **Median (nilai tengah)**
    - Median massa penguin `jantan` atau `Male` species `Adelie` sekitar **4000 gram**
    - sedangkan massa penguin `betina` atau `Female` species `Adelie` sekitar **3400 gram**
